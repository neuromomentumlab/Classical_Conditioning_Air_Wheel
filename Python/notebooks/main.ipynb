# AIR Wheel Facial and Periocular Dynamics Analysis

This notebook serves as the main analysis roadmap for the AIR wheel facial/periocular dynamics project.  
The actual analysis is divided across several notebooks, each responsible for one processing stage.  
To reproduce the analysis from scratch, run the notebooks in the sequence listed below.

The general workflow is:

1. Build and save the data structure
2. Process encoder/behavior signals
3. Validate LED/air/tone event timing
4. Extract event-aligned behavioral epochs
5. Extract facial and periocular motion features
6. Build combined analysis tables
7. Run statistical analyses and generate figures

```python
from IPython.display import display, Markdown

notebooks = [
    ("build_save_view_data_struct.ipynb", "Build `cc_data`, view the data structure, and save project context"),
    ("process_encoder_signal.ipynb", "Process DAQ/MAT files and save encoder/event variables to H5"),
    ("qc_encoder_signal.ipynb", "Check the encoder signals for quality and manually curate if necessary"),
    ("extract_LED_roi_signal.ipynb", "Extract LED ROI signals from videos for synchronization"),
    ("data_exploration.ipynb", "Explore processed data and inspect sessions"),
    ("ami.ipynb", "Compute AMI and event-related locomotor metrics"),
    ("main_runner.ipynb", "Optional runner notebook for pipeline execution"),
    ("test.ipynb", "Temporary testing/debugging notebook"),
]

md = "# AIR Wheel Analysis Notebook Roadmap\n\n"
md += "Run the notebooks in the following sequence.\n\n"
md += "| Step | Notebook | Purpose |\n"
md += "|---|---|---|\n"

for i, (nb, purpose) in enumerate(notebooks, start=1):
    md += f"| {i} | [{nb}](./{nb}) | {purpose} |\n"

display(Markdown(md))
```

## Notebook sequence

### 1. `build_save_view_data_struct.ipynb`

**Purpose:**  
Builds the main data dictionary from the raw data directory. This notebook identifies animals, recording dates, behavioral phases, DAQ files, and associated video files.

**Main inputs:**
- Raw data directory
- Project configuration paths

**Main outputs:**
- `cc_data`
- `data_root`
- `pdata_root`
- Saved project context file for loading in later notebooks

**Run this notebook first.**

### 2. `process_encoder_signal.ipynb`

**Purpose:**  
Processes the raw DAQ/MAT files and extracts wheel encoder and event-timing variables.

**Main outputs for each animal/session:**
- `t`
- `air_raw`
- `air_bin`
- `Air_r`
- `Air_f`
- `encoderCount`
- `dist_net_cm`
- `dist_path_cm`
- `speed_net_cms`
- `speed_path_cms`
- `trial_forward_cm`

These variables are saved into the processed H5 structure for later analysis.

### 3. `extract_LED_roi_signal.ipynb`

**Purpose:**  
Extracts LED intensity from the behavioral video streams to support video-to-behavior synchronization.

**Main outputs:**
- LED intensity signal from each video
- LED-derived event timing
- Video-event alignment information
